## Crear barcos (TERMINADA)

- 3 barcos de eslora 2.
- 2 de eslora 3.
- 1 eslora 4.

```python
def crear_barco(eslora):
    barco = []
    pos_ini = (np.random.randint(10), np.random.randint(10))
    barco.append(pos_ini)
    dir = np.random.choice(("V", "H"))
    p = 1
    row, col = pos_ini

    if dir == "V" and (row + eslora) <= 10:
        for pos in range(eslora - 1):
            barco.append((pos_ini[0] + p, pos_ini[1]))
            p = p + 1
    elif dir == "H" and (col + eslora) <= 10:
        for pos in range(eslora - 1):
            barco.append((pos_ini[0], pos_ini[1] + p))
            p = p + 1   
    else:
        print("calculando barco de nuevo...")
        barco = crear_barco(eslora)

    return barco
```

```python
print(crear_barco(2))
print(crear_barco(4))
```

## Crear flota (TERMINADA)

```python
def crear_flota():
    esloras_flota = [4, 3, 3, 2, 2, 2]
    flota = []
    for eslora in esloras_flota:
        flota.append(crear_barco(eslora))

    lista_posiciones = []
    for barco in flota:
        for pos in barco:
            lista_posiciones.append(pos)

    if len(lista_posiciones) != len(set(lista_posiciones)):
        flota = crear_flota()
    
    return flota
```

## Colocar barcos(TERMINADA)

```python
tablero = crear_tableros()
flota = crear_flota()
```

```python
def colocar_barcos(flota, tablero):
    for barco in flota:
        #print("BARCO: ", barco)
        for pos in barco:
            #print("POS:", pos)
            tablero[pos[0], pos[1]] = "O"

    return tablero
```

## Funcion turnos

```python
def turnos(t_barcos_rival, t_disparos_rival, t_barcos_jugador, t_disparos_jugador):
    turno_jugador = True
    ganador = ""

while True:
    if turno_jugador == True:
        print("Turno jugador")
        fila = int(input("Fila: "))
        col = int(input("Columna: "))
        t_disparos_jugador = disparar(t_barcos_rival, t_disparos_jugador, fila, col)
        turno_jugador = False
        pintar()

    else:
        print("Turno rival")
        fila = np.random.randint(10)
        col = np.random.randint(10)
        t_disparos_rival = disparar(t_barcos_jugador, t_disparos_rival, fila, col)
        turno_jugador = True 
        pintar()

    if "O" not in t_barcos_rival:
        ganador = "jugador"
        break
    if "O" not in t_barcos_jugador:
        ganador = "rival"
        break 
    
return ganador
```
